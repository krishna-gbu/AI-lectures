## SymPy Examples — Chapter 7

```python
import sympy as sp
sp.init_printing()

# ── 7.2 SVD: AᵀA → σᵢ² → vᵢ → uᵢ ──────────────────────────────────
A = sp.Matrix([[3,0],[4,5]])
AtA = A.T * A
print("AᵀA ="); sp.pprint(AtA)

# Right singular vectors = eigenvectors of AᵀA
eig = sorted(AtA.eigenvects(), key=lambda e: -e[0])
sigmas = [sp.sqrt(val) for val, _, _ in eig]
V_cols = [vec[0]/vec[0].norm() for _, _, vecs in eig for vec in vecs]
V = V_cols[0].row_join(V_cols[1])
print("\nσ₁ =", sp.simplify(sigmas[0]), "  σ₂ =", sp.simplify(sigmas[1]))
print("V ="); sp.pprint(sp.simplify(V))

# Left singular vectors: uᵢ = Avᵢ/σᵢ
U_cols = [sp.simplify(A*V_cols[i]/sigmas[i]) for i in range(2)]
U = U_cols[0].row_join(U_cols[1])
print("U ="); sp.pprint(sp.simplify(U))

Sigma = sp.diag(*sigmas)
print("\nUΣVᵀ == A:", sp.simplify(U*Sigma*V.T) == A)

# Pseudoinverse A⁺ = VΣ⁺Uᵀ
Sigma_plus = sp.diag(*[1/s for s in sigmas])
A_plus = sp.simplify(V * Sigma_plus * U.T)
print("AA⁺A == A:", sp.simplify(A*A_plus*A) == A)

# Rank-1 approximation (keep only σ₁)
rank1 = sp.simplify(sigmas[0] * U_cols[0] * V_cols[0].T)
print("\nRank-1 approximation:"); sp.pprint(rank1)
```
