## SymPy Examples — Chapter 2

```python
import sympy as sp
sp.init_printing()

# ── 2.2 Elimination step by step ─────────────────────────────────────
A = sp.Matrix([[2,4,-2],[4,9,-3],[-2,-3,7]])
b = sp.Matrix([2, 8, 10])
Ab = A.row_join(b)
print("rref([A|b]):"); sp.pprint(Ab.rref()[0])
print("Solution x:", A.solve(b).T)    # (-1, 2, 2)

# ── 2.5 Inverse via Gauss-Jordan ─────────────────────────────────────
A2 = sp.Matrix([[2,1],[5,3]])
print("\ndet(A2) =", A2.det())         # 1
print("A2⁻¹ ="); sp.pprint(A2.inv())  # [[3,-1],[-5,2]]
print("A2 * A2⁻¹ ="); sp.pprint(A2 * A2.inv())   # identity

# ── 2.6 A = LU factorisation ─────────────────────────────────────────
A3 = sp.Matrix([[2,1,1],[4,3,3],[8,7,9]])
L, U, _ = A3.LUdecomposition()
print("\nL ="); sp.pprint(L)
print("U ="); sp.pprint(U)
print("L*U == A3:", L*U == A3)

b3 = sp.Matrix([1, 3, 7])
print("Solution via LU:", A3.solve(b3).T)   # (0, 1, 0)

# ── 2.7 Transpose rules ──────────────────────────────────────────────
A4 = sp.Matrix([[1,2],[3,4],[5,6]])
B4 = sp.Matrix([[1,0],[0,1]])
print("\n(AB)ᵀ == BᵀAᵀ:", (A4*B4).T == B4.T * A4.T)   # True
S = A4.T * A4
print("AᵀA symmetric:", S == S.T)    # always True

# LDLᵀ for S = [[2,1],[1,2]]
S2 = sp.Matrix([[2,1],[1,2]])
L2, D2, _ = S2.LDLdecomposition()
print("\nL ="); sp.pprint(L2)
print("D ="); sp.pprint(D2)
print("LDLᵀ == S:", L2*D2*L2.T == S2)
```
