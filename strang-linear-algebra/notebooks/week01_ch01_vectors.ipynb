## SymPy Examples — Chapter 1

Yahan sab kuch purely symbolic hai. NumPy nahi, sirf SymPy. Har cell run karo aur output dekho.

```python
import sympy as sp
sp.init_printing()

# ── 1.1 Linear Combinations ──────────────────────────────────────────
c, d = sp.symbols('c d')
v = sp.Matrix([1, 1, 0])
w = sp.Matrix([0, 1, 1])
print("cv + dw =", c*v + d*w)          # (c, c+d, d)

# Worked Ex 1.1A: normal n ⊥ v and w
n = sp.Matrix([1, -1, 1])
print("v·n =", v.dot(n), "  w·n =", w.dot(n))   # both 0

# Difference matrix: (1,4,9) → (1,3,5)
A = sp.Matrix([[1,0,0],[-1,1,0],[0,-1,1]])
x = sp.Matrix([1, 4, 9])
print("Ax (differences) =", (A*x).T)             # (1,3,5) odd numbers
print("A⁻¹ (sums) ="); sp.pprint(A.inv())         # lower triangular of 1s

# ── 1.2 Dot products and angles ───────────────────────────────────────
v2 = sp.Matrix([1, 0])
w2 = sp.Matrix([1, 1])
angle = sp.acos(v2.dot(w2) / (v2.norm() * w2.norm()))
print("\nangle between (1,0) and (1,1) =", sp.deg(angle), "°")  # 45°

# Perpendicularity check: (3,4) ⊥ (4,-3)?
a, b = sp.Matrix([3,4]), sp.Matrix([4,-3])
print("(3,4)·(4,-3) =", a.dot(b))    # 0 → perpendicular!

# ── 1.3 Invertible vs singular matrix ────────────────────────────────
C = sp.Matrix([[1,0,-1],[-1,1,0],[0,-1,1]])   # cyclic matrix
print("\nCyclic C * (1,1,1) =", (C * sp.Matrix([1,1,1])).T)  # (0,0,0)
print("det(C) =", C.det())            # 0 — singular!
```
