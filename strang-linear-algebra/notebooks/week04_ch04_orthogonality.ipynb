## SymPy Examples — Chapter 4

```python
import sympy as sp
sp.init_printing()

# ── 4.1 Row space ⊥ Nullspace ────────────────────────────────────────
A = sp.Matrix([[1,3,4],[5,2,7]])
x = sp.Matrix([1,1,-1])
print("Ax =", (A*x).T)    # (0,0) — x is in nullspace
# x ⊥ every row of A
print("row1·x =", A.row(0).dot(x), "  row2·x =", A.row(1).dot(x))

# ── 4.2 Projection onto a line ───────────────────────────────────────
a = sp.Matrix([1,2,2])
b = sp.Matrix([1,1,1])
xhat = a.dot(b) / a.dot(a)          # scalar
p = xhat * a
e = b - p
print("\nx̂ =", xhat, "  p =", p.T)
print("e·a =", e.dot(a))            # 0 → perpendicular

P = (a * a.T) / a.dot(a)            # projection matrix
print("\nP² == P:", sp.simplify(P**2 - P) == sp.zeros(3))
print("Pᵀ == P:", P.T == P)

# ── 4.3 Normal equations: best fit line ──────────────────────────────
# Points (0,6),(1,0),(2,0)
A2 = sp.Matrix([[1,0],[1,1],[1,2]])
b2 = sp.Matrix([6, 0, 0])
xhat2 = (A2.T*A2).solve(A2.T*b2)
print("\nBest line C,D:", xhat2.T)   # C=5, D=-3
e2 = b2 - A2*xhat2
print("Errors e:", e2.T, "  A²ᵀe:", (A2.T*e2).T)   # e ⊥ cols

# ── 4.4 Gram-Schmidt ─────────────────────────────────────────────────
a1, a2, a3 = sp.Matrix([1,0,0]), sp.Matrix([1,1,0]), sp.Matrix([1,1,1])
q1 = a1 / a1.norm()
B  = a2 - a2.dot(q1)*q1;  q2 = B/B.norm()
C  = a3 - a3.dot(q1)*q1 - a3.dot(q2)*q2;  q3 = C/C.norm()
Q  = q1.row_join(q2).row_join(q3)
print("\nQᵀQ ="); sp.pprint(sp.simplify(Q.T*Q))    # identity
# QR factorisation
R = sp.Matrix([[q1.dot(a1), q1.dot(a2), q1.dot(a3)],
               [0,          q2.dot(a2), q2.dot(a3)],
               [0,          0,          q3.dot(a3)]])
print("A == QR:", sp.simplify(Q*R) == sp.Matrix([a1,a2,a3]).T)
```
