## SymPy Examples — Chapter 6

```python
import sympy as sp
sp.init_printing()

# ── 6.1 Eigenvalues from det(A-λI)=0 ────────────────────────────────
lam = sp.Symbol('lambda')
A = sp.Matrix([[1,2],[2,1]])
char_poly = sp.expand((A - lam*sp.eye(2)).det())
print("Characteristic poly:", char_poly)         # λ²-2λ-3
evals = sp.solve(char_poly, lam)
print("Eigenvalues:", evals)                      # [-1, 3]
print("Trace =", A.trace(), "= Σλ:", sum(evals))
print("det   =", A.det(),   "= Πλ:", evals[0]*evals[1])
print("\nEigenvects:"); [sp.pprint(v) for _,_,vs in A.eigenvects() for v in vs]

# ── 6.2 Diagonalisation A = XΛX⁻¹ ───────────────────────────────────
X, Lam = A.diagonalize()
print("\nX ="); sp.pprint(X)
print("Λ ="); sp.pprint(Lam)
print("XΛX⁻¹ == A:", sp.simplify(X*Lam*X.inv()) == A)
print("A^10 ="); sp.pprint(sp.simplify(X * Lam**10 * X.inv()))

# Fibonacci
F = sp.Matrix([[1,1],[1,0]])
print("\nF^10 ="); sp.pprint(F**10)
print("F_10 =", (F**10)[0,1])     # 55

# ── 6.4 Symmetric matrix ─────────────────────────────────────────────
S = sp.Matrix([[2,1],[1,2]])
print("\nS eigenvalues:", list(S.eigenvals().keys()))   # 1 and 3 (real)
Qs, Ls = S.diagonalize()
q1 = Qs[:,0] / Qs[:,0].norm()
q2 = Qs[:,1] / Qs[:,1].norm()
Q = q1.row_join(q2)
print("QᵀQ ="); sp.pprint(sp.simplify(Q.T*Q))          # identity

# ── 6.5 Positive definite ────────────────────────────────────────────
S2 = sp.Matrix([[2,1],[1,3]])
x1, x2 = sp.symbols('x1 x2')
energy = (sp.Matrix([x1,x2]).T * S2 * sp.Matrix([x1,x2]))[0]
print("\nxᵀSx =", sp.expand(energy))    # 2x1²+2x1x2+3x2² > 0 always
print("All eigenvalues > 0:", all(v > 0 for v in S2.eigenvals()))
print("All pivots > 0:", all(S2.cholesky()[i,i]**2 > 0
                              for i in range(2)))
```
