## SymPy Examples — Chapter 9

```python
import sympy as sp
sp.init_printing()

# ── 9.1 Euler formula ────────────────────────────────────────────────
print("e^(iπ)    =", sp.exp(sp.I*sp.pi))          # -1
print("e^(iπ/2)  =", sp.exp(sp.I*sp.pi/2))        # i
print("e^(0)     =", sp.exp(0))                    # 1
theta = sp.Symbol('theta', real=True)
mag_sq = sp.simplify(sp.exp(sp.I*theta) * sp.exp(-sp.I*theta))
print("|e^(iθ)|² =", mag_sq)                       # 1

# 4th roots of unity
roots = [sp.exp(2*sp.pi*sp.I*k/4) for k in range(4)]
simplified = [sp.simplify(r) for r in roots]
print("\n4th roots:", simplified)                   # 1, i, -1, -i
print("Sum of roots:", sp.simplify(sum(roots)))     # 0

# ── 9.2 Hermitian matrix: eigenvalues real ───────────────────────────
H = sp.Matrix([[2, 1+sp.I],[1-sp.I, 3]])
print("\nH == H̄ᵀ:", H == H.H)                      # True
evals = list(H.eigenvals().keys())
print("Eigenvalues:", evals)
print("All real:", all(v.is_real for v in evals))  # True

# ── 9.3 DFT matrix F₄ ────────────────────────────────────────────────
w = sp.exp(2*sp.pi*sp.I/4)                          # = i
F4 = sp.Matrix([[w**(j*k) for k in range(4)] for j in range(4)])
print("\nF4:"); sp.pprint(F4)
prod = sp.simplify(F4.H * F4)
print("F̄ᵀF4 ="); sp.pprint(prod)                   # 4I
```
