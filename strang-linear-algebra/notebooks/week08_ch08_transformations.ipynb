## SymPy Examples — Chapter 8

```python
import sympy as sp
sp.init_printing()

theta = sp.Symbol('theta', real=True)

# ── 8.1 Standard 2D transformations ─────────────────────────────────
R = sp.Matrix([[sp.cos(theta),-sp.sin(theta)],
               [sp.sin(theta), sp.cos(theta)]])
R90 = R.subs(theta, sp.pi/2)
print("R(90°)*(1,0) =", (R90 * sp.Matrix([1,0])).T)    # (0,1)
print("R(90°)*(0,1) =", (R90 * sp.Matrix([0,1])).T)    # (-1,0)

Ref = sp.Matrix([[1,0],[0,-1]])   # reflect over x-axis
print("Reflect (1,1):", (Ref * sp.Matrix([1,1])).T)    # (1,-1)

Proj = sp.Matrix([[1,0],[0,0]])   # project onto x-axis
print("Project (3,4):", (Proj * sp.Matrix([3,4])).T)   # (3,0)

# ── 8.2 Basis change: B = M⁻¹AM ─────────────────────────────────────
A = sp.Matrix([[1,2],[0,3]])
X, D = A.diagonalize()            # M = X (eigenvectors)
print("\nA in eigenvector basis D ="); sp.pprint(D)
print("M⁻¹AM == D:", sp.simplify(X.inv()*A*X) == D)
print("Same eigenvalues:", sorted(A.eigenvals()) == sorted(D.eigenvals()))

# ── 8.3 Jordan form (defective matrix) ───────────────────────────────
B = sp.Matrix([[2,1],[0,2]])    # repeated eigenvalue λ=2
print("\nB eigenvects:", B.eigenvects())   # only ONE eigenvector
print("B^5 ="); sp.pprint(B**5)

K = B - 2*sp.eye(2)             # nilpotent part
print("K =", K.T, "  K² =", (K**2).T)    # K²=0
```
