## SymPy Examples — Chapter 5

```python
import sympy as sp
sp.init_printing()

a, b, c, d, t = sp.symbols('a b c d t')
A2 = sp.Matrix([[a,b],[c,d]])

# ── 5.1 All 10 rules verified symbolically ───────────────────────────
print("Rule 1  det(I)    =", sp.eye(2).det())
print("Rule 2  row swap  =", sp.Matrix([[c,d],[a,b]]).det(), " vs -(ad-bc):", -(a*d-b*c))
print("Rule 3a scale row =", sp.expand(sp.Matrix([[t*a,t*b],[c,d]]).det()))
print("Rule 4  equal rows=", sp.Matrix([[a,b],[a,b]]).det())
print("Rule 6  zero row  =", sp.Matrix([[0,0],[c,d]]).det())
print("Rule 9  det(AB)   =", sp.simplify(A2.det() - (A2*sp.eye(2)).det()))  # 0
print("Rule 10 det(Aᵀ)   =", sp.simplify(A2.det() - A2.T.det()))            # 0

# Triangular: det = product of diagonal
U = sp.Matrix([[2,4,-2],[0,1,2],[0,0,4]])
print("\nTriangular det =", U.det(), "= 2×1×4 =", 2*1*4)

# ── 5.1 Worked examples ──────────────────────────────────────────────
M1 = sp.Matrix([[1,1,0],[1,1,1],[0,1,1]])
M2 = sp.Matrix([[1,2,3],[4,5,6],[7,8,9]])
M3 = sp.Matrix([[2,-1],[1,3]])
print("\ndet(M1)=",M1.det(), " det(M2)=",M2.det(), " det(M3)=",M3.det())

# ── 5.2 Cofactor expansion along row 1 ───────────────────────────────
A3 = sp.Matrix([[1,2,3],[4,5,6],[7,8,10]])
C11 =  sp.Matrix([[5,6],[8,10]]).det()
C12 = -sp.Matrix([[4,6],[7,10]]).det()
C13 =  sp.Matrix([[4,5],[7,8]]).det()
manual = 1*C11 + 2*C12 + 3*C13
print("\nCofactor expansion =", manual, "  SymPy =", A3.det())

# ── 5.3 Cramer's rule ────────────────────────────────────────────────
A4 = sp.Matrix([[2,1],[1,3]])
b4 = sp.Matrix([5,10])
A4c1 = A4.copy(); A4c1[:,0] = b4
A4c2 = A4.copy(); A4c2[:,1] = b4
x1 = A4c1.det() / A4.det()
x2 = A4c2.det() / A4.det()
print("\nCramer x =", x1, x2)
print("Verify:", (A4 * sp.Matrix([x1,x2])).T)
```
