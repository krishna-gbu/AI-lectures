## SymPy Examples — Chapter 12

```python
import sympy as sp
sp.init_printing()

# ── 12.1 Mean and variance ────────────────────────────────────────────
vals  = [1, 2, 3]
probs = [sp.Rational(1,6), sp.Rational(2,6), sp.Rational(3,6)]
mu  = sum(v*p for v,p in zip(vals,probs))
var = sum((v-mu)**2*p for v,p in zip(vals,probs))
print("E[X]   =", mu)
print("Var(X) =", sp.simplify(var))
print("E[X²]-μ² =", sp.simplify(sum(v**2*p for v,p in zip(vals,probs)) - mu**2))

# ── 12.2 Covariance matrix ────────────────────────────────────────────
X = sp.Matrix([[1,2],[3,1],[2,3]])        # 3 samples, 2 features
mu_vec = sum(X.row(i) for i in range(3)) / 3
Xc = sp.Matrix([X.row(i) - mu_vec.T for i in range(3)])
V = (Xc.T * Xc) / 2                      # divide by n-1=2
print("\nCovariance V:"); sp.pprint(V)
print("V symmetric:", V == V.T)
print("Eigenvalues:", list(V.eigenvals().keys()))

# Independence ↔ rank 1
P_joint = sp.Matrix([[sp.Rational(1,6),sp.Rational(1,6)],
                      [sp.Rational(2,6),sp.Rational(2,6)]])
print("\nJoint prob rank:", P_joint.rank())   # 1 → independent

# ── 12.3 Weighted least squares ───────────────────────────────────────
A = sp.Matrix([[1,0],[1,1],[1,2]])
b = sp.Matrix([6, 0, 0])
W = sp.diag(4, 1, 1)              # high weight on first obs
xhat_w = (A.T*W*A).solve(A.T*W*b)
xhat_o = (A.T*A).solve(A.T*b)
print("\nWeighted LS (C,D):", xhat_w.T)
print("Ordinary LS (C,D):", xhat_o.T)
```
