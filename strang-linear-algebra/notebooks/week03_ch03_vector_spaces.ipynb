## SymPy Examples — Chapter 3

```python
import sympy as sp
sp.init_printing()

# ── 3.2 Nullspace ─────────────────────────────────────────────────────
A = sp.Matrix([[1,2,3,1],[2,4,7,2],[3,6,10,3]])
R, pivots = A.rref()
print("rref(A):"); sp.pprint(R)
print("Pivot columns:", pivots)     # (0, 2)

ns = A.nullspace()
print("\nNullspace basis:")
for v in ns: sp.pprint(v)
print("A * ns[0] =", (A * ns[0]).T)   # (0,0,0)
print("A * ns[1] =", (A * ns[1]).T)

# ── 3.3 Complete solution xp + xn ────────────────────────────────────
b = sp.Matrix([1, 2, 3])
# Particular solution: free variables = 0
Ab = A.row_join(b)
Rb, _ = Ab.rref()
print("\n[R|d]:"); sp.pprint(Rb)
# Read off xp from pivot rows
xp = sp.Matrix([Rb[0,4], 0, Rb[1,4], 0])
print("Particular xp:", xp.T)
print("Verify A*xp == b:", A*xp == b)

# ── 3.4 Independence test ────────────────────────────────────────────
v1 = sp.Matrix([1,2,3])
v2 = sp.Matrix([2,4,6])     # dependent: = 2*v1
v3 = sp.Matrix([1,0,1])
M = v1.row_join(v2).row_join(v3)
print("\ndet[v1,v2,v3] =", M.det())    # 0 → dependent
print("rank =", M.rank())              # 2

M2 = v1.row_join(v3)
print("rank[v1,v3] =", M2.rank())     # 2 → independent

# ── 3.5 Four subspaces summary ───────────────────────────────────────
print("\nC(A) basis (pivot cols of A):")
sp.pprint(A[:, list(pivots)])
print("C(Aᵀ) basis (nonzero rows of R):")
sp.pprint(R[:A.rank(), :])
print("dim C(A) =", A.rank(), "  dim N(A) =", A.shape[1]-A.rank())
```
