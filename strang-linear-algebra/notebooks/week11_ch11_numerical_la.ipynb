## SymPy Examples — Chapter 11

```python
import sympy as sp
sp.init_printing()

# ── 11.1 Tiny pivot → catastrophic error ─────────────────────────────
eps = sp.Rational(1, 10000)
A_bad = sp.Matrix([[eps, 1],[1, 1]])
b1 = sp.Matrix([1, 2])
x_bad = A_bad.solve(b1)
print("Tiny pivot solution:", x_bad.T)

A_good = sp.Matrix([[1,1],[eps,1]])   # row swap (partial pivoting)
x_good = A_good.solve(sp.Matrix([2,1]))
print("After row swap  :", x_good.T)  # same answer, numerically safer

# ── 11.2 Condition number κ = σ₁/σₙ ─────────────────────────────────
A = sp.Matrix([[4,1],[1,3]])
sv = sorted(A.singular_values(), reverse=True)
kappa = sp.simplify(sv[0]/sv[1])
print("\nSingular values:", [sp.simplify(s) for s in sv])
print("Condition number κ =", kappa)

# Small Δb → how large Δx?
b_exact  = sp.Matrix([1, 0])
b_perturb = sp.Matrix([1 + sp.Rational(1,100), 0])
dx = A.solve(b_perturb) - A.solve(b_exact)
print("δb = 1% →  δx:", sp.simplify(dx).T)

# ── 11.3 Jacobi iteration (6 steps) ─────────────────────────────────
x = sp.Matrix([0, 0])
b2 = sp.Matrix([1, 2])
print("\nJacobi for [4,1;1,3]x=[1,2]:")
for k in range(6):
    x = sp.Matrix([(b2[0] - x[1])/4, (b2[1] - x[0])/3])
    print(f"  step {k+1}: {x.T}")
print("Exact:", A.solve(b2).T)
```
